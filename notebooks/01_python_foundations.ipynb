## Collections and set comparisons

        Sets are useful when comparing planned and completed identifiers.

```python
planned_wells = {"W-01", "W-02", "W-03", "W-04"}
completed_wells = {"W-01", "W-03", "W-04", "W-05"}
print("Missing:", sorted(planned_wells - completed_wells))
print("Unexpected:", sorted(completed_wells - planned_wells))

well = {"well_id": "W-04", "pressure_mpa": 25.2, "bubble_point_mpa": 18.4}
status = "above bubble point" if well["pressure_mpa"] > well["bubble_point_mpa"] else "at or below bubble point"
print(well["well_id"], status)
```
