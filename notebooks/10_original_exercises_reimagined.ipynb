```python
from petroleum_python.petrophysics import core_bulk_volume_m3, oil_in_place_stb, porosity
from petroleum_python.units import atm_to_psi, psi_to_mpa

core_volume = core_bulk_volume_m3(diameter_m=0.0381, length_m=0.0762)
phi = porosity(pore_volume=24.0, bulk_volume=100.0)
pressures_mpa = psi_to_mpa([1000, 1200, 1400, 1600, 1800])
ooip_stb = oil_in_place_stb(10, 50, 0.25, 0.20, 1.20)
print(f"Core volume: {core_volume:.8f} m³")
print(f"Porosity: {phi:.1%}")
print("Pressure (MPa):", pressures_mpa.round(3))
print(f"1 atm = {atm_to_psi(1):.4f} psi")
print(f"OOIP: {ooip_stb:,.0f} STB")
```

```python
def compare_identifiers(reference, observed):
    reference, observed = set(reference), set(observed)
    return {
        "missing": sorted(reference - observed),
        "additional": sorted(observed - reference),
        "common": sorted(reference & observed),
    }

compare_identifiers(["A", "B", "C", "D"], ["C", "D", "E"])
```

```python
def pressure_relative_to_bubble_point(pressure_mpa, bubble_point_mpa, tolerance_mpa=0.05):
    if pressure_mpa <= 0 or bubble_point_mpa <= 0:
        raise ValueError("pressures must be positive")
    difference = pressure_mpa - bubble_point_mpa
    if abs(difference) <= tolerance_mpa:
        return "near bubble point"
    return "above bubble point" if difference > 0 else "below bubble point"

for pressure in [16.0, 18.4, 22.0]:
    print(pressure, pressure_relative_to_bubble_point(pressure, 18.4))
```
